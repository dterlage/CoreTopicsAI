```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import (accuracy_score, confusion_matrix, classification_report)
from google.colab import files
```

```python
# Importing dataframes
baseline_1 = pd.read_csv("/content/ravdess_baseline_predictions.csv")
baseline_2 = pd.read_csv("/content/ravdess_baseline_predictions_2.csv")
finetuned_1 = pd.read_csv("/content/ravdess_finetuned_predictions.csv")
finetuned_2 = pd.read_csv("/content/ravdess_finetuned_predictions_2.csv")
```

```python
emotion_labels = ["neutral", "calm", "happy", "sad", "angry", "fearful", "disgust", "surprised"]
```

```python

y_true_b1 = baseline_1["actual_emotion"]
y_pred_b1 = baseline_1["predicted_emotion"]

y_true_b2 = baseline_2["actual_emotion"]
y_pred_b2 = baseline_2["predicted_emotion"]

y_true_f1 = finetuned_1["actual_emotion"]
y_pred_f1 = finetuned_1["predicted_emotion"]

y_true_f2 = finetuned_2["actual_emotion"]
y_pred_f2 = finetuned_2["predicted_emotion"]

accuracy_b1 = accuracy_score(y_true_b1, y_pred_b1)
accuracy_b2 = accuracy_score(y_true_b2, y_pred_b2)
accuracy_f1 = accuracy_score(y_true_f1, y_pred_f1)
accuracy_f2 = accuracy_score(y_true_f2, y_pred_f2)

print(f"Accuracy Baseline prompt 1: {accuracy_b1:.4f}")
print(f"Accuracy Baseline prompt 2: {accuracy_b2:.4f}")
print(f"Accuracy Finetuned prompt 1: {accuracy_f1:.4f}")
print(f"Accuracy Finetuned prompt 2: {accuracy_f2:.4f}")
```

```python
# Confusion matrix
cm = confusion_matrix(
    y_true_b1,
    y_pred_b1,
    labels=emotion_labels
    )

# Plot heatmap
plt.figure(figsize=(10, 8))

sns.heatmap(
    cm,
    annot=True,
    cmap="Blues",
    xticklabels=emotion_labels,
    yticklabels=emotion_labels
)

plt.title("Confusion Matrix Baseline prompt 1", fontsize=16)
plt.xlabel("Predicted Emotion", fontsize=12)
plt.ylabel("Actual Emotion", fontsize=12)

plt.tight_layout()
plt.savefig("cm_baseline_1.png", bbox_inches='tight')
files.download("cm_baseline_1.png")
plt.show()
```

```python
# Confusion matrix
cm = confusion_matrix(
    y_true_b2,
    y_pred_b2,
    labels=emotion_labels
)

# Plot heatmap
plt.figure(figsize=(10, 8))

sns.heatmap(
    cm,
    annot=True,
    cmap="Blues",
    xticklabels=emotion_labels,
    yticklabels=emotion_labels
)

plt.title("Confusion Matrix Baseline prompt 2", fontsize=16)
plt.xlabel("Predicted Emotion", fontsize=12)
plt.ylabel("Actual Emotion", fontsize=12)

plt.tight_layout()
plt.savefig("cm_baseline_2.png", bbox_inches='tight')
files.download("cm_baseline_2.png")
plt.show()
```

```python
# Confusion matrix
cm = confusion_matrix(
    y_true_f1,
    y_pred_f1,
    labels=emotion_labels
    )

# Plot heatmap
plt.figure(figsize=(10, 8))

sns.heatmap(
    cm,
    annot=True,
    cmap="Blues",
    xticklabels=emotion_labels,
    yticklabels=emotion_labels
)

plt.title("Confusion Matrix Fine-tuned model prompt 1", fontsize=16)
plt.xlabel("Predicted Emotion", fontsize=12)
plt.ylabel("Actual Emotion", fontsize=12)

plt.tight_layout()
plt.savefig("cm_finetuned_1.png", bbox_inches='tight')
files.download("cm_finetuned_1.png")
plt.show()
```

```python
# Confusion matrix
cm = confusion_matrix(
    y_true_f2,
    y_pred_f2,
    labels=emotion_labels
    )

# Plot heatmap
plt.figure(figsize=(10, 8))

sns.heatmap(
    cm,
    annot=True,
    cmap="Blues",
    xticklabels=emotion_labels,
    yticklabels=emotion_labels
)

plt.title("Confusion Matrix Fine-tuned prompt 2", fontsize=16)
plt.xlabel("Predicted Emotion", fontsize=12)
plt.ylabel("Actual Emotion", fontsize=12)

plt.tight_layout()
plt.savefig("cm_finetuned_2.png", bbox_inches='tight')
files.download("cm_finetuned_2.png")
plt.show()
```

```python
# Precison, Recall, F1
# Baseline prompt 1
report = classification_report(
    y_true_b1,
    y_pred_b1,
    labels=emotion_labels,
    target_names=emotion_labels,
    zero_division=0,
    output_dict=True
)

report_df = pd.DataFrame(report).transpose()

display(report_df.round(3))
```

```python
# Precison, Recall, F1
# Baseline prompt 2
report = classification_report(
    y_true_b2,
    y_pred_b2,
    labels=emotion_labels,
    target_names=emotion_labels,
    zero_division=0,
    output_dict=True
)

report_df = pd.DataFrame(report).transpose()

display(report_df.round(3))
```

```python
# Precison, Recall, F1-score
# Finetuned prompt 1
report = classification_report(
    y_true_f1,
    y_pred_f1,
    labels=emotion_labels,
    target_names=emotion_labels,
    zero_division=0,
    output_dict=True
)

report_df = pd.DataFrame(report).transpose()

display(report_df.round(3))
```
